```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
N = 5000
x = np.random.uniform(-1, 1, N)
y = np.random.uniform(-1, 1, N)

inside = (x**2 + y**2) <= 1.0
pi_est = 4 * np.cumsum(inside) / np.arange(1, N + 1)

fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Left Panel: Random Sampling
axs[0].scatter(x[inside], y[inside], color='royalblue', s=3, label='Inside Circle')
axs[0].scatter(x[~inside], y[~inside], color='crimson', s=3, label='Outside Circle')
axs[0].set_title(f"Monte Carlo Sampling (N={N}, Est. Pi={pi_est[-1]:.4f})")
axs[0].legend(loc="upper right")

# Right Panel: Convergence Plot
axs[1].plot(pi_est, color='darkgreen', lw=1.5, label='Estimated Pi')
axs[1].axhline(np.pi, color='black', linestyle='--', label='True Pi (3.14159)')
axs[1].set_ylim(2.8, 3.5)
axs[1].set_title("Law of Large Numbers Convergence")
axs[1].set_xlabel("Number of Iterations (N)")
axs[1].legend()

plt.tight_layout()
plt.savefig("monte_carlo.png", dpi=300)
plt.show()
```
